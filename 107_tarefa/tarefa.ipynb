```python
import control as ctl
import numpy as np
import matplotlib.pyplot as plt

Num = D = [1,1.5,2]
Den = [1,0,0]

sys = ctl.tf(Num,Den)

polos =ctl.poles(sys)
zeros = ctl.zeros(sys)

print("Polos:",polos)
print("Zeros:",zeros)

ctl.pzmap(sys, grid=True, title="Mapa de Polos e Zeros")
plt.show()
```

```python
N = 5

n = np.arange(N)

x = []
y = np.zeros(N)

for i in range(N):
    if i == 0:
        x.append(1)
    else:
        x.append(0)

for i in range(N):
    if i == 0:
        y[i] = x[i]

    elif i == 1:
        y[i] = x[i] - 0.25 * y[i-1]

    else:
        y[i] = x[i] - 0.25 * y[i-1] - 0.125 * y[i-2]

print(y)

# Plotar
plt.figure(figsize=(10, 5))

plt.stem(n, y,
         linefmt='b-',
         markerfmt='bo',
         label='Entrada y[n] - Impulso Unitário')

plt.title('Resposta do Sistema ao Impulso Unitário')
plt.xlabel('Amostras n')
plt.ylabel('Amplitude')
plt.legend()
plt.grid(True)
plt.show()
```
