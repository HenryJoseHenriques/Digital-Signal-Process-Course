```python
import numpy as np
import matplotlib.pyplot as plt

N = 8000
n = []
Xd = []

for i in range(0,8):
    n.append(i)
    Xd.append((np.sin(2*np.pi*1000*i/N) + 0.5*np.sin(2*np.pi*2000*i/N + (3*np.pi/4)) ))

# Plotar
plt.figure(figsize=(10, 5))

plt.stem(n, Xd,
         linefmt='b-',
         markerfmt='bo',
         label='Entrada Xd(t)')

plt.title('Resposta do Sistema ao Xd(t)')
plt.xlabel('Amostras n')
plt.ylabel('Amplitude')
plt.legend()
plt.grid(True)
plt.show()

print(Xd)
```

```python
Xm = []

for i in range(N):
    total = 0
    for j in range(N):
        angle = 2 * np.pi * i * j / N
        # Use 1j for the imaginary unit instead of np.i
        total += Xd[j] * (np.cos(angle) - 1j * np.sin(angle))
    Xm.append(total)

Xm = np.array(Xm)
```
