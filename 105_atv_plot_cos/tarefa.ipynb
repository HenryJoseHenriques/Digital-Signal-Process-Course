Tarefa: 

Dado 

x[n] = 2d[n+2]  + d[n+1] + d[n] + d[n-1] + 2d[n-2] 

Obtenha sua DFT: 
Plote |X(e^(jw))| = Somatorio(-inf,+inf) x[n].e(-jwn)  

```python
import numpy as np
import matplotlib.pyplot as plt

tam = 5
x = []
n = []

# Mantendo a sua lógica exata de criação do sinal
for i in range(-tam, tam):
    n.append(i)
    if i == -2 or i == 2:
        x.append(2)
    elif -1 <= i <= 1:
        x.append(1)
    else:
        x.append(0)

# Converter listas para arrays do NumPy
x = np.array(x)
n = np.array(n)

# Limpar figuras anteriores abertas
plt.close('all')

# Vetor de frequências w de -pi a pi com passo de pi/100
w = np.arange(-1 * np.pi, 1 * np.pi + np.pi/100, np.pi/100)

# --- CORREÇÃO DA DTFT ---
# Usamos [:, None] e [None, :] para multiplicar a matriz (n x w) de forma vetorizada.
# Isso calcula o somatório da DTFT: X(w) = soma( x[n] * e^(-j * w * n) )
X = np.sum(x[:, None] * np.exp(-1j * n[:, None] * w[None, :]), axis=0)

# Cálculo do Módulo (Magnitude) e Fase do sinal resultante completo X
Mod_X = np.abs(X)
Fase_X = np.angle(X)

# --- Plotando o Módulo e a Fase ---
plt.figure(figsize=(8, 6))

# Subplot 1: Módulo
plt.subplot(2, 1, 1)
plt.plot(w, Mod_X)
plt.title('Modulo de X')
plt.xlabel('w')
plt.grid(True)

# Subplot 2: Fase
plt.subplot(2, 1, 2)
plt.plot(w, Fase_X)
plt.title('Fase de X')
plt.xlabel('w')
plt.grid(True)

# Ajustar espaçamento entre os gráficos e exibir
plt.tight_layout()
plt.show()
```
