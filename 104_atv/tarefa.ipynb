# Média Movel para Impulso Unitário

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal

x = []
n = []

for i in range(0, 10):
    n.append(i)

    if i == 0:
        x.append(1)
    else:
        x.append(0)

# Filtro média móvel de janela M = 8
M = 8

b = np.ones(M) / M
a = [1]

# Processar o sinal
y = signal.lfilter(b, a, x)

# Plotar
plt.figure(figsize=(10, 5))

plt.stem(n, x,
         linefmt='b-',
         markerfmt='bo',
         label='Função de Impulso Unitário')

plt.stem(n, y,
         linefmt='r-',
         markerfmt='ro',
         label=f'Saída do Sistema $y[n]$ (Móvel M={M})')
plt.title('Resposta do Filtro Média Móvel ao Impulso Unitário')
plt.xlabel('Amostras n')
plt.ylabel('Amplitude')
plt.legend()
plt.grid(True)
plt.show()
```

# Degrau unitário
