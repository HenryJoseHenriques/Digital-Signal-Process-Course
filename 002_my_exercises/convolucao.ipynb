```python
import numpy as np
import matplotlib.pyplot as plt

# --------------------------------------------------
# Definição de x[n]
# --------------------------------------------------

# Valores de x[n]
x = np.array([5, 2, 3], dtype=float)
nx = np.arange(-2, 1)

# --------------------------------------------------
# Definição de h[n]
# --------------------------------------------------

# Valores de h[n]
h = np.array([2, 4], dtype=float)
nh = np.arange(-2, 0)

# --------------------------------------------------
# Cálculo da convolução
# --------------------------------------------------

y = np.convolve(x, h)

# O primeiro índice da saída é a soma dos
# primeiros índices de x[n] e h[n]
inicio_y = nx[0] + nh[0]

# O último índice é a soma dos últimos índices
fim_y = nx[-1] + nh[-1]

# Índices da saída
ny = np.arange(inicio_y, fim_y + 1)

# --------------------------------------------------
# Exibição dos resultados
# --------------------------------------------------

print("Índices de x[n]:", nx)
print("Valores de x[n]:", x)

print("\nÍndices de h[n]:", nh)
print("Valores de h[n]:", h)

print("\nÍndices de y[n]:", ny)
print("Valores de y[n]:", y)

# --------------------------------------------------
# Gráficos
# --------------------------------------------------

plt.figure(figsize=(10, 9))

# Gráfico de x[n]
plt.subplot(3, 1, 1)
plt.stem(nx, x, basefmt="k-")
plt.title("Sinal de entrada x[n]")
plt.xlabel("n")
plt.ylabel("Amplitude")
plt.axhline(0, color="black", linewidth=0.8)
plt.axvline(0, color="black", linewidth=0.8)
plt.xticks(nx)
plt.grid(True, linestyle="--", alpha=0.6)

# Gráfico de h[n]
plt.subplot(3, 1, 2)
plt.stem(nh, h, basefmt="k-")
plt.title("Resposta ao impulso h[n]")
plt.xlabel("n")
plt.ylabel("Amplitude")
plt.axhline(0, color="black", linewidth=0.8)
plt.axvline(0, color="black", linewidth=0.8)
plt.xticks(nh)
plt.grid(True, linestyle="--", alpha=0.6)

# Gráfico de y[n]
plt.subplot(3, 1, 3)
plt.stem(ny, y, basefmt="k-")
plt.title("Convolução y[n] = x[n] * h[n]")
plt.xlabel("n")
plt.ylabel("Amplitude")
plt.axhline(0, color="black", linewidth=0.8)
plt.axvline(0, color="black", linewidth=0.8)
plt.xticks(ny)
plt.grid(True, linestyle="--", alpha=0.6)

plt.tight_layout()
plt.show()
```
